```python
import pandas as pd

df = pd.read_csv(
    "./data/delivery_performance_cleaned.csv",
    parse_dates=["order_date", "expected_delivery_date", "actual_delivery_date"]
)
df.info()
```

### **EDA and KPIs**

```python
# datset shape
df.shape

dims_count = {
    "Orders Count": df["order_id"].nunique(),
    "Warehouses Count": df["warehouse_id"].nunique()
}
dims_count
```

### **Orders Distribution by Warehouse**

```python
# orders distribution by warehouse
orders_by_warehouse = (
    df
        .groupby("warehouse_id")
        .agg(
            orders_count = ("order_id", "nunique")
        )
        .sort_values("orders_count", ascending = False)
)
orders_by_warehouse["percentage"] = (orders_by_warehouse["orders_count"] * 100 / orders_by_warehouse["orders_count"].sum()).round(2)
orders_by_warehouse
```

```python
# orders segmentation
df["delivery_status"] = (
    df["actual_delivery_date"] <= df["expected_delivery_date"]
).map(
    {
        True: "On-Time",
        False: "Late"
    }
)
df.head()
```

### **delivery_status orders distribution**

```python
orders_by_delivery_status = (
    df
        .groupby("delivery_status")
        .agg(
            orders_count = ("order_id", "nunique")
        )
)
orders_by_delivery_status
```
