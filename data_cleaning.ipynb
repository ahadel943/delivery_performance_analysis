```python
import pandas as pd

df = pd.read_csv(
    "./data/delivery_performance.csv",
    parse_dates=["order_date", "expected_delivery_date", "actual_delivery_date"]
)

# Remove rows with missing warehouse_id
analytical_df = df.dropna(subset=["warehouse_id"]).copy()

# Remove exact duplicate rows
analytical_df = analytical_df.drop_duplicates().copy()

# Remove duplicated order IDs
analytical_df = analytical_df.drop_duplicates(
    subset="order_id",
    keep=False
).copy()
```

### **Analytical data quality check**

```python
print("Original rows:", len(df))
print("Analytical rows:", len(analytical_df))

print("\nMissing values:")
print(analytical_df.isnull().sum())

print("\nExact duplicates:")
print(analytical_df.duplicated().sum())

print("\nDuplicate order IDs:")
print(analytical_df["order_id"].duplicated().sum())
```

```python
analytical_df.to_csv("./data/delivery_performance_cleaned.csv")
```
